# TruthLens: Multi-Modal Robustness & Adversarial Evaluation

This notebook benchmarks the robustness of the TruthLens multi-modal detectors under realistic
social-media perturbations, out-of-distribution cross-dataset transfer, and white-box adversarial attacks.

- **Image**: JPEG compression (90/70/50), downscale-resize (0.5x), Gaussian blur, screenshot-style quantization, and social-media compression.
- **Audio**: MP3 lossy codec compression (128/64 kbps), additive noise (SNR 20/10 dB), and telephonic phone-band (300-3400 Hz) filtering.
- **Text**: Paraphrasing, synonym swaps, simulated keyboard typos, and human/AI paragraph mixing.
- **Adversarial**: White-box FGSM and PGD attacks on the image model, and empirical adversarial training defense.
- **Generalization**: Cross-dataset transfer evaluation and retention metrics.
- **Publication Figures**: Generates high-resolution IEEE publication charts in `paper/figures/`.

## 1. Environment Setup and Imports

```python
from pathlib import Path

import pandas as pd

from ml.eval.robustness import (
    compute_cross_dataset_generalization,
    plot_adversarial_curves,
    plot_audio_robustness,
    plot_combined_robustness_grid,
    plot_cross_dataset_drop,
    plot_image_robustness,
    plot_text_robustness,
    run_adversarial_attack_benchmark,
    run_audio_perturbation_benchmark,
    run_image_perturbation_benchmark,
    run_text_perturbation_benchmark,
)

FIG_DIR = Path("../paper/figures")
FIG_DIR.mkdir(parents=True, exist_ok=True)
print("TruthLens Robustness Suite initialized successfully.")
```

## 2. Image Perturbation Benchmark (JPEG, Resize, Blur, Social Media)

Evaluates ConvNeXt-Tiny with DCT frequency branch against lossy re-encoding and optics perturbations.

```python
df_img = run_image_perturbation_benchmark()
plot_image_robustness(df_img, FIG_DIR / "robustness_image_perturbations.png")
display(df_img) if 'display' in globals() else print(df_img)
```

## 3. Audio Perturbation Benchmark (Codecs, Noise, Telephony)

Evaluates LCNN + LFCC against MP3 128/64 kbps, pink/white noise, and ITU-T G.712 phone-band pass.

```python
df_aud = run_audio_perturbation_benchmark()
plot_audio_robustness(df_aud, FIG_DIR / "robustness_audio_perturbations.png")
display(df_aud) if 'display' in globals() else print(df_aud)
```

## 4. Text Perturbation Benchmark (Stylometry, Typos, Mixing)

Evaluates ensemble meta-classifier detection rate under paraphrasing, synonym swaps, and human sentence intermixing.

```python
df_text = run_text_perturbation_benchmark()
plot_text_robustness(df_text, FIG_DIR / "robustness_text_perturbations.png")
display(df_text) if 'display' in globals() else print(df_text)
```

## 5. Adversarial Attacks (FGSM & PGD) and Empirical Adversarial Training

Evaluates standard baseline model vs adversarially hardened model under increasing perturbation epsilon.

```python
df_adv = run_adversarial_attack_benchmark()
plot_adversarial_curves(df_adv, FIG_DIR / "adversarial_fgsm_pgd_curve.png")
display(df_adv) if 'display' in globals() else print(df_adv)
```

## 6. Cross-Dataset Generalization Analysis

Measures performance drop delta between in-domain training distributions and novel out-of-domain evaluation benchmarks.

```python
gen_results = compute_cross_dataset_generalization()
plot_cross_dataset_drop(gen_results, FIG_DIR / "cross_dataset_generalization_drop.png")
df_gen = pd.DataFrame([vars(r) for r in gen_results])
display(df_gen) if 'display' in globals() else print(df_gen)
```

## 7. Master Publication 4-Panel Grid Figure

```python
grid_path = FIG_DIR / "combined_robustness_grid.png"
plot_combined_robustness_grid(df_img, df_aud, df_text, df_adv, grid_path)
print(f"Master publication grid saved: {grid_path}")
```
